```python
import pandas as pd
```

```python

df = pd.read_csv('RNPDNO-22-08-2023.csv',sep=',',encoding='latin-1')
```

```python
mask = df.apply(lambda row: row.astype(str).str.contains(r'^ELIMINADO', regex=True).any(), axis=1)
df_limpio = df[~mask]
```

```python
df_limpio.to_csv('RNPDNO-22-08-2023-limpio.csv', index=False, encoding='latin-1')
```
